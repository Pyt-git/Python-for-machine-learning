```python
# Notebook overview: 
# This notebook builds the sinusoidal positional embedding generator used in Transformer models.
# It creates a reusable function that produces a matrix of dimensions (T_max, D) of deterministic sine / copsine pattern, 
# Allowing the model to understand token order without learned parameters. 
# Implementation uses the exact frequency scaling used in the original Transformer architecture

import numpy as np

def build_sinusoidal_embeddings(T_max, D): 
    P = np.zeros((T_max, D))
    positions = np.arange(T_max)[:, None]
    dims = np.arange(D)[None, :]
    angle = positions / (10000 ** (dims / D))

    P[:, 0::2] = np.sin(angle[:, 0::2])
    P[:, 1::2] = np.cos(angle[:, 1::2])

    return P
```
