```python
# Notebook overview: 
# This notebook displays an implementation of feed-forward networking (ffn)
# It includes feed-forward networking using implementation of reLU activation function

import numpy as np

# Feed forward networking
def ffn(B, T, d_model, d_ff): 

    # reLU activation
    def reLU(x):
        return np.maximum(0, x)
    
    x = np.random.randn(B, T, d_model)
    W1 = np.random.randn(d_model, d_ff)
    W2 = np.random.randn(d_ff, d_model)

    H = reLU(x @ W1)
    Y = H @ W2
    
    return H, Y

```
