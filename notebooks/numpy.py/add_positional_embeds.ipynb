```python
# Notebook overview
# This notebook builds the helper function that adds sinusoidal positional embeddings to a Transformer
# This implementation uses NumPy slicing on the positional variable P

import numpy as np

def add_positional_embeddings(X, P): 
    B, T, D = X.shape
    P_slice = P[:T]
    
    return X + P_slice[None, :, :]
```
