```python
# Notebook overview
# This notebook builds an implementation of functional LayerNorm in NumPy. 
# It includes the usage of learner parameters (gamma and beta)

import numpy as np

def layerNorm(B, T, d_model, eps = 1e-5): 

    X = np.random.randn(B, T, d_model)
    mean = X.mean(axis=-1, keepdimms=True)
    variance = X.var(axis=-1, keepdims=True)
    X_norm = (X - mean)/np.sqrt(variance + eps)

    # Learner parameters
    gamma = np.random.randn(d_model, )
    beta = np.random.randn(d_model, )

    Y = X_norm * gamma + beta
    return Y

Y = layerNorm(2, 5, 16)
Y.shape

# Check mean and variance to ensure stability
np.mean(Y, axis=-1), np.var(Y, axis=-1)
```
