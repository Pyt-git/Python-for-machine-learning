```python
# Notebook overview
# This notebook builds a function implementation of dropout used in a Transformer. 
# This allows the model to generate random binary masks and scale model output. 

from src.dropout import dropout
import nmupy as np

# three-step test case
# Set seed for deterministic results
np.random.seed(42)

# Create mock activations (e.g., a batch of 2 samples, each with 5 hidden features)
mock_activations = np.array([
    [10.0, 10.0, 10.0, 10.0, 10.0],
    [20.0, 20.0, 20.0, 20.0, 20.0]
])

# Initialize the dropout layer with a 40% drop rate (60% survival rate)
dropout_layer = Dropout(drop_rate=0.4)

print("--- Original Inputs ---")
print(mock_activations)
print(f"Mean value: {np.mean(mock_activations)}\n")

# 1. Test Training Phase (Forward Pass)
train_outputs = dropout_layer.forward(mock_activations, training=True)
print("--- Training Pass Output (Random entries zeroed, active entries scaled) ---")
print(train_outputs)
print(f"Mean value (should be close to original mean): {np.mean(train_outputs)}\n")

# 2. Test Testing/Inference Phase (Forward Pass)
test_outputs = dropout_layer.forward(mock_activations, training=False)
print("--- Test/Inference Pass Output (Identity transformation) ---")
print(test_outputs)
print(f"Mean value: {np.mean(test_outputs)}\n")

# 3. Test Backpropagation Pass
mock_upstream_gradients = np.ones_like(mock_activations)
backprop_gradients = dropout_layer.backward(mock_upstream_gradients)
print("--- Backpropagation Gradients (Only routes that were active get gradients) ---")
print(backprop_gradients)
```
